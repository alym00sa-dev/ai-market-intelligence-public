# Add secrets to the `frontier_labs` scope

1. Fill in the key values in the `secrets` dict below (uncomment the ones you need).
2. **Run the cell.**
3. **Delete the values** from the cell afterward (the secret is stored; the only exposed copy is the cell text).

Reference (from `.env`): `AA_API_KEY` = the `aa_…` key · `LLM_STATS_KEY` = the `sk_…` key · `ANTHROPIC_API_KEY` = the `sk-ant-…` key.

```python
from databricks.sdk import WorkspaceClient
w = WorkspaceClient()
SCOPE = "frontier_labs"

# create the scope once (safe to re-run)
try:
    w.secrets.create_scope(scope=SCOPE)
    print(f"created scope: {SCOPE}")
except Exception as e:
    print(f"scope already exists (or): {e}")

# fill in only the keys you want to add/update, then run — then clear the values
secrets = {
    # "AA_API_KEY":        "aa_...",
    # "LLM_STATS_KEY":     "sk_...",
    # "ANTHROPIC_API_KEY": "sk-ant-...",
}
for k, v in secrets.items():
    w.secrets.put_secret(scope=SCOPE, key=k, string_value=v)
    print(f"stored: {k}")

print("keys now in scope:", [s.key for s in w.secrets.list_secrets(SCOPE)])
```
