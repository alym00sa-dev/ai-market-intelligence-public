# Add a secret (reusable)

Type a **Key Name** and **Key Value** into the two boxes that appear at the top after running cell 2, then run cell 3. No value is hard-coded in the notebook.

1. Run **cell 1** once (creates the `frontier_labs` scope if missing).
2. Run **cell 2** once (creates the two input boxes).
3. Fill the boxes → run **cell 3** to store the secret. Repeat 3 for each key.
4. Run **cell 4** to verify (shows key *names* only, never values).

_Note: the value is briefly visible in the widget bar while set — cell 3 clears it after. For zero exposure, use the Databricks CLI instead._

```python
# Cell 1 — create scope (run once; safe to re-run)
from databricks.sdk import WorkspaceClient
w = WorkspaceClient()
SCOPE = "frontier_labs"
try:
    w.secrets.create_scope(scope=SCOPE)
    print(f"created scope: {SCOPE}")
except Exception as e:
    print(f"scope already exists (or): {e}")
```

```python
# Cell 2 — create the input boxes (run once). Then fill them at the top of the notebook.
dbutils.widgets.text("key_name", "")
dbutils.widgets.text("key_value", "")
```

```python
# Cell 3 — store the secret (run AFTER filling both boxes). Re-run for each new key.
k = dbutils.widgets.get("key_name").strip()
v = dbutils.widgets.get("key_value")
assert k, "fill the 'key_name' box at the top first"
assert v, "fill the 'key_value' box at the top first"
w.secrets.put_secret(scope=SCOPE, key=k, string_value=v)
print(f"stored '{k}' in scope '{SCOPE}'")
# clear the value box so it isn't left visible
dbutils.widgets.remove("key_value")
dbutils.widgets.text("key_value", "")   # recreate empty for the next key
```

```python
# Cell 4 — verify (names only, no values)
print([s.key for s in w.secrets.list_secrets(scope=SCOPE)])
```
